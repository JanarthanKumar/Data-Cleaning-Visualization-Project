```python
import pandas as pd
pd.set_option('display.max_columns',None)
```

```python
data=pd.read_excel(r"C:\Users\Janar\OneDrive\Documents\data_sales.xlsx")
data.head(44)
```

```python
print("\nMissing values count:")
print(data.isnull().sum())
```

```python
data['Age']=data['Age'].fillna(data['Age'].median())
data['Salary']=data['Salary'].fillna(data['Salary'].mean())
data=data.dropna(subset=['Join_Date'])
print(data)
```

```python
data['Join_Date']=pd.to_datetime(data['Join_Date'], format='mixed',errors='coerce')
data['Join_Date']=data['Join_Date'].dt.strftime('%Y-%m-%d')
print(data)
```

```python
data=data.apply(lambda x: x.str.strip()if x.dtype =="object" else x) 
```

```python
data=data.apply(lambda x: x.str.lower()if x.dtype=="object" else x)
print(data)
```

```python
data=data.drop_duplicates(subset=['Name','Age',])
print(data)
```

```python
print(data.duplicated().sum())
```

```python
Q1=data['Salary'].quantile(0.25)
Q3=data['Salary'].quantile(0.75)

IQR=Q3-Q1

data=data[(data['Salary']>=Q1-1.5*IQR)&(data['Salary']<=Q3+1.5*IQR)]
print(data)
```

```python
data['Join_Date']=pd.to_datetime(data['Join_Date'], format='mixed',errors='coerce')
data['Join_Date']=data['Join_Date'].dt.strftime('%d/%m/%Y')
print(data)
```

```python
data=pd.get_dummies(data,columns=['Department'],drop_first=True)
binary_columns=[col for col in data.columns if col.startswith('Department_')]
data[binary_columns]=data[binary_columns].astype(int)
print(data)
```

```python
data.to_csv("cleaned_data.csv",index=False)
```

```python
import matplotlib.pyplot as plt

data=pd.read_csv("cleaned_data.csv")

print(data.head())
```

```python
department_counts = data[['Department_HR','Department_IT']].sum()

department_counts.plot(kind='bar', color=['yellow', 'green'], figsize=(8, 6))
plt.title("Department Distribution")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(ticks=[0, 1], labels=["HR", "IT"], rotation=0)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.hist(data['Salary'], bins=10, color='orange', edgecolor='black')
plt.title("Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
plt.plot(data['Age'], data['Salary'], marker='o', linestyle='-', color='blue')
plt.title("Salary vs. Age")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 6))
colors = ['blue' if dept == 1 else 'green' for dept in data['Department_HR']]
plt.scatter(data['Salary'], data['Age'], c=colors, alpha=0.7, edgecolor='black')
plt.title("Salary vs. Age by Department (HR in Blue & IT in Green)")
plt.xlabel("Salary")
plt.ylabel("Age")
plt.grid(True)
plt.show()
```
